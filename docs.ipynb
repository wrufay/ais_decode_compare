## Section 1: Introduction

This section provided a lot of context regarding Automatic Identification System (AIS) and its use cases, including *enhancing safety of life at sea, safety and efficiency of naviation, as well as protection of the marine environment.*

Differences between terrestrial AIS datasets (like the Canadian Coast Guard we are working with here) and satellite AIS dataset (were working with pre-decoded satellite data last week):
- *Terrestrial* AIS datasets have a **higher temporal resolution** - meaning **more frequent data points over time**/a finer time scale. This is because land-based stations are able to constantly recieve signals as opposed to satellite signals which depend on the **frequency of orbital passes**. This is one advantage of terrestrial datasets.
- However, a disadvantage of terrestrial AIS datasets is that signals can easily be lost if the **vessel moves behind a land mass or another vessel.** The height of the **reciever** (land) and the height of the **transponder** (boat) both play a role in the distance at which an AIS will be recieved. 

### Additional notes
- AIS data is used to address **conservation** issues
- AIS data is ideal for both **historical** and **real-time** analytics
- AIS data is transmitted and recorded in coded messages, must be first decoded which can be a challenge considering messages sizes can range from 10 to 40 million for just one day.

### Different types of AIS messages (new, unfamiliar with this)
- *Location* related message types are **1, 2, 3, 5, 18**
- *Safety* related message types are *12, 14*, which are not able to be decoded using the previous packages developed by DFO, one written in Fortran and one in R. Being able to decode safety related messages was one of the purposes of developing this package, as well as improving the efficiency and speed through using Python.
