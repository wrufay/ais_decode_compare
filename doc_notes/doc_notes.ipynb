## Section 1

### AIS context
- Variety of use cases, enhancing safety of sea life, navigation efficiency, marine environment protection

### Differences between terrestrial and satellite AIS datasets
- *Terrestrial* AIS datasets: **higher temporal resolution** - aka **more frequent data points over time** (advantage)
- Reason: land-based stations are able to constantly recieve signals, satellite signals depend on **frequency of orbital passes**

- Disadvantage: signals can easily be lost if the **vessel moves behind a land mass or another vessel.** 
- Reciever (land) and transponder (boat) height both matter for the distance the AIS will be recieved

### Additional notes on AIS data
- Used to address **conservation** issues
- Ideal for both **historical** and **real-time** analytics
- Transmitted and recorded in coded messages, must be first decoded
- Messages in one day can range from 10 to 40 million

### Different types of AIS messages (new, unfamiliar with this)
- *Location* related message types are **1, 2, 3, 5, 18**
- *Safety* related message types are **12, 14**; not able to be decoded using the previous packages developed
